***Data Exploration & Cleaning***

```python
import pandas as pd

base_path = "/content/sentnob_data/SentNoB Dataset"

train_df = pd.read_csv(f"{base_path}/Train.csv")
val_df = pd.read_csv(f"{base_path}/Val.csv")
test_df = pd.read_csv(f"{base_path}/Test.csv")

print("Train shape :", train_df.shape)
print("Val shape :", val_df.shape)
print("Test shape :", test_df.shape)

print("\nColumns :", train_df.columns.tolist())
print("\nFirst 5 rows:")
train_df.head()
```

```python
print("Label distribution (Train) :")
print(train_df["Label"].value_counts())
print("\nLabel distribution (%) :")
print(train_df["Label"].value_counts(normalize=True) * 100)

print("\nUnique labels :", train_df["Label"].unique())

print("\nSample texts per label:")
for label in train_df["Label"].unique():
    sample = train_df[train_df["Label"] == label]["Data"].iloc[0]
    print(f"\nLabel {label} : {sample}")
```

```python
import re

def clean_bangla_text(text):
    text = str(text)

    text = re.sub(r"http\S+|www\S+", "", text)

    text = re.sub(r"[^\u0980-\u09FF\u0020-\u007E]", " ", text)

    text = re.sub(r"([!?.,])\1+", r"\1", text)

    text = re.sub(r"\s+", " ", text).strip()
    return text

train_df["clean_text"] = train_df["Data"].apply(clean_bangla_text)
val_df["clean_text"] = val_df["Data"].apply(clean_bangla_text)
test_df["clean_text"] = test_df["Data"].apply(clean_bangla_text)

print("Empty after cleaning (train):", (train_df["clean_text"].str.strip() == "").sum())

for i in range(3):
    print("\nBefore:", train_df["Data"].iloc[i])
    print("After :", train_df["clean_text"].iloc[i])
```

```python
print("Nulls in train :", train_df.isnull().sum().sum())
print("Duplicates in train :", train_df.duplicated(subset=["Data"]).sum())

train_df["word_count"] = train_df["clean_text"].apply(lambda x: len(x.split()))
print("\nWord count stats:")
print(train_df["word_count"].describe())
```

```python
print("Before dedup:", train_df.shape)
train_df = train_df.drop_duplicates(subset=["Data"]).reset_index(drop=True)
print("After dedup:", train_df.shape)

print("Val duplicates:", val_df.duplicated(subset=["Data"]).sum())
print("Test duplicates:", test_df.duplicated(subset=["Data"]).sum())
```

***Tokenization***

```python
val_df = val_df.drop_duplicates(subset=["Data"]).reset_index(drop=True)
test_df = test_df.drop_duplicates(subset=["Data"]).reset_index(drop=True)

print("Val shape:", val_df.shape)
print("Test shape:", test_df.shape)
```

```python
from datasets import Dataset

MAX_LENGTH = 64

def tokenize_function(examples):
    return tokenizer(examples["clean_text"], truncation=True, padding="max_length", max_length=MAX_LENGTH)

train_dataset = Dataset.from_pandas(train_df[["clean_text", "Label"]].rename(columns={"Label": "labels"}))
val_dataset = Dataset.from_pandas(val_df[["clean_text", "Label"]].rename(columns={"Label": "labels"}))
test_dataset = Dataset.from_pandas(test_df[["clean_text", "Label"]].rename(columns={"Label": "labels"}))

train_dataset = train_dataset.map(tokenize_function, batched=True)
val_dataset = val_dataset.map(tokenize_function, batched=True)
test_dataset = test_dataset.map(tokenize_function, batched=True)

train_dataset = train_dataset.remove_columns(["clean_text"])
val_dataset = val_dataset.remove_columns(["clean_text"])
test_dataset = test_dataset.remove_columns(["clean_text"])

train_dataset.set_format("torch")
val_dataset.set_format("torch")
test_dataset.set_format("torch")

print(train_dataset)
print("\nSample:", train_dataset[0])
```

***MOdel Training***

```python
from transformers import AutoModelForSequenceClassification

num_labels = 3

model = AutoModelForSequenceClassification.from_pretrained(model_name, num_labels=num_labels)

print(model.config)
```

```python
from sklearn.utils.class_weight import compute_class_weight
import torch

labels_array = train_df["Label"].values
class_weights = compute_class_weight(class_weight="balanced", classes=np.unique(labels_array), y=labels_array)
class_weights = torch.tensor(class_weights, dtype=torch.float)
print("Class weights:", class_weights)
```

```python
import torch.nn as nn
from transformers import Trainer
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

class WeightedTrainer(Trainer):
    def compute_loss(self, model, inputs, return_outputs=False, num_items_in_batch=None):
        labels = inputs.pop("labels")
        outputs = model(**inputs)
        logits = outputs.logits
        loss_fct = nn.CrossEntropyLoss(weight=class_weights.to(logits.device))
        loss = loss_fct(logits, labels)
        return (loss, outputs) if return_outputs else loss

def compute_metrics(eval_pred):
    logits, labels = eval_pred
    preds = np.argmax(logits, axis=1)
    precision, recall, f1, _ = precision_recall_fscore_support(labels, preds, average="weighted")
    acc = accuracy_score(labels, preds)
    return {"accuracy": acc, "f1": f1, "precision": precision, "recall": recall}
```

```python
from transformers import TrainingArguments

training_args = TrainingArguments(output_dir="/content results", eval_strategy="epoch", save_strategy="epoch", learning_rate=2e-5, per_device_train_batch_size=16, per_device_eval_batch_size=32, num_train_epochs=4, weight_decay=0.01, logging_dir="/content logs", logging_steps=50, load_best_model_at_end=True, metric_for_best_model="f1", save_total_limit=2, report_to="none")
```

```python
import datasets
datasets.config.TORCHVISION_AVAILABLE = False
```

```python
trainer = WeightedTrainer(model=model, args=training_args, train_dataset=train_dataset, eval_dataset=val_dataset, compute_metrics=compute_metrics)

trainer.train()
```

***Evaluation***

```python
test_results = trainer.evaluate(test_dataset)
print("Test set results:")
for key, value in test_results.items():
    print(f"{key}: {value}")
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import confusion_matrix, classification_report
import seaborn as sns

predictions = trainer.predict(test_dataset)
preds = np.argmax(predictions.predictions, axis=1)
true_labels = predictions.label_ids

# Classification report
label_names = ["Neutral (0)", "Positive (1)", "Negative (2)"]
print(classification_report(true_labels, preds, target_names=label_names))

# Confusion matrix
cm = confusion_matrix(true_labels, preds)
plt.figure(figsize=(6, 5))
sns.heatmap(cm, annot=True, fmt="d", cmap="Blues", xticklabels=label_names, yticklabels=label_names)
plt.xlabel("Predicted")
plt.ylabel("Actual")
plt.title("Confusion Matrix - Test Set")
plt.show()
```

***Error Analysis***

```python
test_df_reset = test_df.reset_index(drop=True)
test_df_reset["predicted"] = preds
test_df_reset["true_label"] = true_labels

# Vul prediction gulo dekhi, specially Neutral (0) niye
errors = test_df_reset[test_df_reset["true_label"] != test_df_reset["predicted"]]
print(f"Total errors: {len(errors)} out of {len(test_df_reset)}")

# Neutral ke onno kisu hisebe vul predict kora cases
neutral_errors = errors[errors["true_label"] == 0]
print(f"\nNeutral misclassified: {len(neutral_errors)}")
print("\nSample misclassified Neutral texts:")
for i in range(min(5, len(neutral_errors))):
    row = neutral_errors.iloc[i]
    print(f"\nText: {row['clean_text']}")
    print(f"True: Neutral(0) | Predicted: {row['predicted']}")

# Onno dik theke- ki onno class ke Neutral bole predict korche
predicted_neutral_wrong = errors[errors["predicted"] == 0]
print(f"\n\nWrongly predicted as Neutral: {len(predicted_neutral_wrong)}")
for i in range(min(5, len(predicted_neutral_wrong))):
    row = predicted_neutral_wrong.iloc[i]
    print(f"\nText: {row['clean_text']}")
    print(f"True: {row['true_label']} | Predicted: Neutral(0)")
```

```python
%%writefile app.py
import streamlit as st
import torch
import numpy as np
import re
from transformers import AutoTokenizer, AutoModelForSequenceClassification

MODEL_PATH = "/content/bangla_sentiment_model"
MAX_LENGTH = 64
LABEL_MAP = {0: "Neutral", 1: "Positive", 2: "Negative"}
LABEL_COLORS = {0: "gray", 1: "green", 2: "red"}

@st.cache_resource
def load_model():
    tokenizer = AutoTokenizer.from_pretrained(MODEL_PATH)
    model = AutoModelForSequenceClassification.from_pretrained(MODEL_PATH)
    model.eval()
    return tokenizer, model

def clean_bangla_text(text):
    text = str(text)
    text = re.sub(r"http\S+|www\S+", "", text)
    text = re.sub(r"[^\u0980-\u09FF\u0020-\u007E]", " ", text)
    text = re.sub(r"([!?.,])\1+", r"\1", text)
    text = re.sub(r"\s+", " ", text).strip()
    return text

def predict(text, tokenizer, model):
    cleaned = clean_bangla_text(text)
    inputs = tokenizer(cleaned, return_tensors="pt", truncation=True, padding="max_length", max_length=MAX_LENGTH)
    with torch.no_grad():
        outputs = model(**inputs)
        probs = torch.softmax(outputs.logits, dim=1).numpy()[0]
    pred_label = int(np.argmax(probs))
    return pred_label, probs

st.set_page_config(page_title="Bangla Sentiment Analysis", page_icon="🇧🇩")
st.title("🇧🇩 Bangla Text Sentiment Analysis")
st.write("Bangla-BERT দিয়ে তৈরি — একটা বাংলা বাক্য লিখুন, sentiment জানুন।")

tokenizer, model = load_model()

sample_sentences = [
    "আজকে আবহাওয়া টা বেশ ভালো লাগছে",
    "এই জিনিসটা একদম ভালো লাগে নাই আমার কাছে",
    "কালকে অফিসে মিটিং আছে সকাল দশটায়"
]

st.write("**Sample sentences:**")
cols = st.columns(3)
selected_sample = None
for i, sent in enumerate(sample_sentences):
    if cols[i].button(sent[:20] + "...", key=f"sample_{i}"):
        selected_sample = sent

text_input = st.text_area("আপনার বাংলা টেক্সট লিখুন:", value=selected_sample if selected_sample else "", height=100)

if st.button("Predict Sentiment", type="primary"):
    if text_input.strip() == "":
        st.warning("অনুগ্রহ করে কিছু টেক্সট লিখুন।")
    else:
        pred_label, probs = predict(text_input, tokenizer, model)
        sentiment = LABEL_MAP[pred_label]
        confidence = probs[pred_label] * 100

        st.markdown(f"### Prediction: **:{LABEL_COLORS[pred_label]}[{sentiment}]**")
        st.write(f"Confidence: {confidence:.2f}%")

        st.write("**Probability breakdown:**")
        for label_id, label_name in LABEL_MAP.items():
            st.write(f"{label_name}: {probs[label_id]*100:.2f}%")
            st.progress(float(probs[label_id]))
```
